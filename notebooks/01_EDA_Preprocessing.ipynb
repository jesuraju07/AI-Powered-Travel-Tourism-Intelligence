```python
cleaned_destinations = destinations.copy()
```

```python
cleaned_destinations["latitude"] = pd.to_numeric(
    cleaned_destinations["latitude"], errors="coerce"
)

cleaned_destinations["longitude"] = pd.to_numeric(
    cleaned_destinations["longitude"], errors="coerce"
)

cleaned_destinations["wikidata_pagerank"] = pd.to_numeric(
    cleaned_destinations["wikidata_pagerank"], errors="coerce"
)
```

Validate geographical coordinates

```python
cleaned_destinations = cleaned_destinations[
    cleaned_destinations["latitude"].between(-90, 90) &
    cleaned_destinations["longitude"].between(-180, 180)
]
```

Tourist Reviews — Cleaning code

Create a copy

```python
cleaned_reviews = reviews.copy()
```

Remove duplicate records

```python
cleaned_reviews = cleaned_reviews.drop_duplicates()

cleaned_reviews = cleaned_reviews.drop_duplicates(
    subset="text"
)
```

Extract prediction labels and scores

```python
import ast

def extract_prediction(value):
    try:
        result = ast.literal_eval(value)

        if isinstance(result, list) and len(result) > 0:
            return result[0]["label"], result[0]["score"]

    except (ValueError, SyntaxError, TypeError, KeyError):
        pass

    return np.nan, np.nan
```

```python
cleaned_reviews[["rating", "prediction_score"]] = (
    cleaned_reviews["prediction"].apply(
        lambda x: pd.Series(extract_prediction(x))
    )
)
```

Convert data types

```python
cleaned_reviews["rating"] = pd.to_numeric(
    cleaned_reviews["rating"], errors="coerce"
)

cleaned_reviews["prediction_score"] = pd.to_numeric(
    cleaned_reviews["prediction_score"], errors="coerce"
)

cleaned_reviews["event_timestamp"] = pd.to_datetime(
    cleaned_reviews["event_timestamp"], errors="coerce", utc=True
)
```

Clean review text

```python
cleaned_reviews["text"] = (
    cleaned_reviews["text"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)
```

Remove invalid records

```python
cleaned_reviews = cleaned_reviews.dropna(
    subset=["text", "rating"]
)

cleaned_reviews = cleaned_reviews[
    (cleaned_reviews["text"] != "") &
    (cleaned_reviews["rating"].between(1, 5)) &
    (cleaned_reviews["prediction_score"].between(0, 1))
]
```

Create sentiment categories

```python
def sentiment_category(rating):

    if rating <= 2:
        return "Negative"

    elif rating == 3:
        return "Neutral"

    else:
        return "Positive"


cleaned_reviews["sentiment"] = cleaned_reviews["rating"].apply(
    sentiment_category
)
```

Select useful columns

```python
cleaned_reviews = cleaned_reviews[
    [
        "text",
        "rating",
        "prediction_score",
        "sentiment",
        "event_timestamp"
    ]
]
```

Final verification

```python
print("Original Shape:", reviews.shape)

print("Cleaned Shape:", cleaned_reviews.shape)

print("\nMissing Values:")
print(cleaned_reviews.isnull().sum())

print("\nDuplicate Rows:")
print(cleaned_reviews.duplicated().sum())

print("\nSentiment Distribution:")
print(cleaned_reviews["sentiment"].value_counts())

print("\nSample Cleaned Reviews:")
display(cleaned_reviews.head(10))
```
